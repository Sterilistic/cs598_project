# Load CheXpert Dataset

This notebook imports the CheXpertDataset class from pyhealth/datasets/chexpert.py and shows how to instantiate it.

Run the cells in order.

```python
from pathlib import Path
import sys
import importlib.util

# Make sure the local PyHealth package is importable from this notebook.
# Notebook location: PyHealth/examples/rexkg/load_dataset.ipynb
# Package root:      PyHealth/
pyhealth_root = Path.cwd().resolve().parents[1]
if str(pyhealth_root) not in sys.path:
    sys.path.insert(0, str(pyhealth_root))

try:
    from pyhealth.datasets.chexpert import CheXpertDataset
except Exception:
    chexpert_file = pyhealth_root / "pyhealth" / "datasets" / "chexpert.py"
    spec = importlib.util.spec_from_file_location("chexpert_local", chexpert_file)
    module = importlib.util.module_from_spec(spec)
    spec.loader.exec_module(module)
    CheXpertDataset = module.CheXpertDataset

print(f"Using PyHealth root: {pyhealth_root}")
print("Imported CheXpertDataset successfully")
```

```python
# Set this to the folder that contains your CheXpert files.
# Required files for best results include:
# - df_chexpert_plus_240401.csv (or another recognized metadata CSV)
# - image files referenced by the CSV path_to_image column
chexpert_root = Path("/path/to/chexpert_root")

dataset = CheXpertDataset(root=str(chexpert_root))
print(dataset)

dataset.stat()
```

```python
# Optional: build the default task and inspect one processed sample
samples = dataset.set_task()
print(f"Number of task samples: {len(samples)}")
if len(samples) > 0:
    print(samples[0])
    
```
